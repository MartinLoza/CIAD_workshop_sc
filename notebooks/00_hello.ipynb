## 1. Are we actually running R?

This must print an R version. If it errors, or prints something about Python,
the runtime is wrong: go to **Runtime → Change runtime type** and choose **R**.

```r
cat(R.version.string, "\n")
cat("platform:", R.version$platform, "\n")
```

## 2. Which machine are we on?

Colab gives you a temporary Linux machine in a Google data centre. Nothing here
runs on your own laptop. This cell reports which Linux it is.

Note how we call a shell command: from R we use `system()`, not `!`.

```r
cat(system("cat /etc/os-release 2>&1", intern = TRUE), sep = "\n")
```

## 3. Where do R packages live?

The workshop notebooks will install packages into this folder. It is wiped when
the runtime shuts down, which is normal and expected.

```r
cat("library paths:\n")
cat(paste0("  ", .libPaths()), sep = "\n")
cat("\npackages already installed:", nrow(installed.packages()), "\n")
```

## 4. Can we draw a plot?

Figures are the point of most of this workshop, so let's confirm one appears.
The `repr.plot` options set the size of every figure in the notebook; we set
them once at the top of every notebook.

```r
options(repr.plot.width = 10, repr.plot.height = 7)

plot(
  iris$Petal.Length, iris$Petal.Width,
  col  = iris$Species,
  pch  = 19,
  xlab = "petal length",
  ylab = "petal width",
  main = "If you can see this, plotting works"
)
legend("topleft", legend = levels(iris$Species), col = 1:3, pch = 19)
```
